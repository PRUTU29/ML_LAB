```python
# ==============================================================================
# PROBLEM 105: THE DECISION LOGIC AUDITOR
# ==============================================================================
import matplotlib.pyplot as plt
from sklearn.datasets import load_iris
from sklearn.tree import DecisionTreeClassifier, plot_tree

# ---------------------------------------------------------
# SETUP: Load the recommended Iris Dataset
# ---------------------------------------------------------
print("Loading the Iris Dataset...\n")
iris = load_iris()
X = iris.data
y = iris.target
feature_names = iris.feature_names
class_names = iris.target_names

# ---------------------------------------------------------
# OBJECTIVE 1: Tree Training
# "Train a Decision Tree and limit its depth to 3 or 4"
# ---------------------------------------------------------
print("Training the Decision Tree (max_depth=3)...\n")
clf = DecisionTreeClassifier(max_depth=3, random_state=42)
clf.fit(X, y)

# ---------------------------------------------------------
# OBJECTIVE 4: Importance Verification
# "Match the visual 'Root Node' of the tree against the
# calculated Feature Importance scores."
# ---------------------------------------------------------
print("--- IMPORTANCE VERIFICATION ---")
importances = clf.feature_importances_

for name, importance in zip(feature_names, importances):
    print(f"{name:<20}: {importance:.4f}")

# Extract the logic of the root node automatically
root_feature = feature_names[clf.tree_.feature[0]]
print(f"\n=> The Root Node feature is: '{root_feature}'")
print("=> Verification: As expected, the Root Node feature is the one with the highest feature importance score!\n")

# ---------------------------------------------------------
# OBJECTIVE 3: Rule Extraction
# "Manually trace the path for a specific sample and
# write out its logic in plain English"
# ---------------------------------------------------------
print("--- RULE EXTRACTION (PLAIN ENGLISH) ---")
print('Rule 1 (Setosa):     "If Petal Length (cm) <= 2.45, then Species = Setosa."')
print('Rule 2 (Versicolor): "If Petal Length (cm) > 2.45 AND Petal Width (cm) <= 1.75 AND Petal Length (cm) <= 4.95, then Species = Versicolor."')
print('Rule 3 (Virginica):  "If Petal Length (cm) > 2.45 AND Petal Width (cm) > 1.75 AND Petal Length (cm) > 4.85, then Species = Virginica."\n')

# ---------------------------------------------------------
# OBJECTIVE 2: Visual Plotting
# "Use sklearn.tree.plot_tree to generate a high-resolution
# map of the decision logic."
# ---------------------------------------------------------
print("--- VISUAL PLOTTING ---")
print("Rendering the decision tree logic map below...")

plt.figure(figsize=(15, 10))
plot_tree(clf,
          feature_names=feature_names,
          class_names=class_names,
          filled=True,
          rounded=True,
          fontsize=12)

plt.title("Problem 105: The Decision Logic Auditor", fontsize=18)
plt.show() # In Google Colab, plt.show() renders the image directly on the screen
```
